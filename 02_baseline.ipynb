```python
from google.colab import drive
drive.mount('/content/drive')

import torch
import pandas as pd
import matplotlib.pyplot as plt

from torch_geometric.datasets import EllipticBitcoinDataset
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report

dataset = EllipticBitcoinDataset(root='/content/drive/MyDrive/data-mining/elliptic')
data = dataset[0]
print(data)
```

用官方自带的 mask 把训练集和测试集抠出来

```python
X_train = data.x[data.train_mask].numpy()
y_train = data.y[data.train_mask].numpy()
X_test = data.x[data.test_mask].numpy()
y_test = data.y[data.test_mask].numpy()

print("训练集节点数：", X_train.shape[0])
print("测试集节点数：", X_test.shape[0])
```

决策树基线
- recall = 猜对的欺诈数 / 真实欺诈总数
- precision = 猜对的欺诈数 / 你一共猜了多少个欺诈
- F1 = 2×precision×recall / (precision+recall)
-

```python
clf = DecisionTreeClassifier(class_weight='balanced', random_state=42)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)
print("决策树结果：")
print(classification_report(y_test, y_pred, target_names=['正常(0)', '欺诈(1)']))
```

SVM 基线

```python
svm = SVC(class_weight='balanced', random_state=42)
svm.fit(X_train, y_train)

y_pred_svm = svm.predict(X_test)
print("SVM结果：")
print(classification_report(y_test, y_pred_svm, target_names=['正常(0)', '欺诈(1)']))
```
